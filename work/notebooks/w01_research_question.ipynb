## 3. Quick look at the data (2-3 real numbers)

Here is a quick look at the starter dataset to prove this problem is worth solving. The data below shows that a massive portion of the inventory is failing, and that simple, single-metric rules (like checking the word count) are not enough to identify the real problems.

```python
# 1. What percentage of the inventory is currently declining?
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)
declining_rate = df["is_declining_label"].mean()

# 2. Does word count easily separate growing from declining pages?
wc_down = df[df["trend_direction"] == "down"]["word_count"].median()
wc_up = df[df["trend_direction"] == "up"]["word_count"].median()

print("=== LANE 2 DATA JUSTIFICATION ===")
print(f"1. Decay Volume: {declining_rate:.1%} of the {len(df):,} pages are declining. The team desperately needs prioritization.")
print(f"2. Signal Complexity: Declining pages have a median word count of {wc_down:.0f}, while growing pages have {wc_up:.0f}. A simple human heuristic isn't enough; we need a proper model.")
```
